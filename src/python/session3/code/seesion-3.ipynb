```python
def factorial(num : int) :
    ''' 
    calculate n! using recursion

    parameters 
    ----------
    num : int

    returns
    -------
    int
    '''
    if num < 0 :
        return ValueError("Factorial is undefiend for negative values")
    if num == 0 or num == 1 :
        return 1
    return factorial(num - 1) * num
```

```python
factorial(7)
```

```python
def is_prime(num) ->bool :
    ''' 
    Chech whether a number is prime 

    paramter
    --------
    num : int

    return
    ------
    bool

    True if prime , otherwise False
    '''
    if num < 2 :
        return False
    for i in range(2, num) :
        if num % i == 0 :
            return False 
    return True
```

```python
is_prime(7)
```

```python
def common_divisors(num1 : int , num2 : int) -> list[int] :
    limit = min(num1, num2)
    divisors = []

    for divisor in range(1, limit + 1) :
        if num1 % divisor == 0 and num2 % divisor == 0 :
            divisors.append(divisor)
    return divisors
```

```python
common_divisors(10, 20)
```

```python
def reverse_words(words : str) -> list[str] :
    '''
    function take words and reverse it in list 

    parameter
    ---------
    words : string 

    return
    ------
    list of string

    '''
    lst = words.split(" ")
    new_list = lst[::-1]
    return new_list
```

```python
reverse_words("Hello Everyone i love everyone here")
```

```python
import random 
import string
def generate_password(pass_length : int = 12) -> str:
    '''
    Function to Generate Password

    parameter
    ---------
    pass_length : int

    return
    ------
    password : string
    
    '''
    if pass_length <= 0 :
        ValueError ('Password length must be greater than 0')
    
    char = (
            string.ascii_letters +
            string.digits +
            string.punctuation
    )

    password = ""

    for _ in range(pass_length) :
        password += random.choice(char)
    return password

 
```

```python
generate_password()
```
