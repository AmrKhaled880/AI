```python
class Solution:
    def __init__(self, nums : list[int], target : int) -> list[int] :
        self.nums = nums
        self.target = target

    def twoSum(self):
        sol_list = []

        for i in range(len(self.nums)) :
            for j in range(i + 1,len(self.nums)) :

                if self.nums[i] + self.nums[j] == self.target :
                    sol_list.append(i)
                    sol_list.append(j) 
                    return sol_list 
```

```python
list1 = [3, 2, 4]
s1 = Solution(list1, 6)
print(s1.twoSum())
```
