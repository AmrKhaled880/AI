```python
class ListNode :
    def __init__(self, val = 0, next = None):
        self.val = val
        self.next = next
        
class Solution :    
    def addTwoNumbers(self, l1: Optional[ListNode], l2: Optional[ListNode]) -> Optional[ListNode]:
        final_list = ListNode()
        carry = 0
        rem = 0

        while l1 != None:
                result = 0
                result = l1.val + l2.val + carry
                carry = 0
                rem = 0
                newnode = ListNode()


                if  result < 10:
                     final_list.val = result
                     final_list.next = newnode
                     final_list = newnode
                     l1 = l1.next
                     l2 = l2.next


                elif result >= 10 :
                     carry = result // 10
                     rem = result % 10
                     final_list.val = rem
                     final_list.next = newnode
                     final_list = newnode
                     l1 = l1.next
                     l2 = l2.next

                     
                 


        
```
