```python
def array_reshape(arr, rows = -1, cols = -1) :
    try :
        arr = np.array(arr)
        return arr.reshape((rows, cols))
    
    except Exception as e:
        raise ValueError('Invalid reshape operations : {e}')
```

```python
vec = [1, 4, 2, 4, 11, 22]
print('test 1(2 * 3)matrix' , array_reshape(vec, 2 ,3))
```
